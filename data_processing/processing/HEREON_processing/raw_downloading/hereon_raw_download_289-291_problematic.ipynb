```python
# Function to get list of CSV file URLs from the webpage
def get_csv_urls(webpage_url):
    response = requests.get(webpage_url)
    response.raise_for_status()
    html_content = response.text

    # Regular expression to find CSV links
    csv_pattern = re.compile(r'href=["\'](.*?\.csv)["\']', re.IGNORECASE)
    csv_urls = csv_pattern.findall(html_content)

    # Convert relative URLs to absolute URLs
    absolute_csv_urls = []
    for url in csv_urls:
        if not url.startswith('http'):
            url = os.path.join(webpage_url, url)
        absolute_csv_urls.append(url)
    
    return absolute_csv_urls
```

```python
for drifter_num in drifter_nums:
    # URL of the webpage to scrape
    webpage_url = 'https://www.socib.es/users/dataprocuser/FaSt-SWOT/hereon_drifters_raw_RT/csv/' + str(drifter_num) + '/'

    # Get the list of CSV URLs
    csv_urls = get_csv_urls(webpage_url)

    dataframes = []
    for url in csv_urls:
        print(url)
        try:
            # Attempt to read the CSV file from the URL
            df = pd.read_csv(url)
        
        except pd.errors.EmptyDataError:
            # Handle the EmptyDataError
            print("EmptyDataError encountered: No columns to parse from file.")
            print("Problematic file: " + url)
            # Create a DataFrame with the current date and time
            
            # Original timestamp string
            original_timestamp = url.split('/')[-1].split('.')[0].split('_')[-1]

            # Parse the original timestamp string into a datetime object
            parsed_timestamp = datetime.strptime(original_timestamp, '%Y%m%dT%H%M%S')

            # Format the datetime object into the desired string format
            formatted_timestamp = parsed_timestamp.strftime('%d-%b-%Y %H:%M:%S')

            # Display the formatted timestamp
            print(formatted_timestamp)
            
            # Assuming you know the column names you want in the DataFrame
            column_names = ['D_number', 'date_UTC', 'Latitude', 'Longitude', 'U_speed_mps', 'U_Dir_deg', 'batteryState'] 

            # Create a DataFrame with one row and assign values to the first two columns and NaN to the rest
            data = [[str(drifter_num), formatted_timestamp] + [np.nan] * (len(column_names) - 2)] ## double check drifter_num format is good!!
            df = pd.DataFrame(data, columns=column_names)

        dataframes.append(df)

    # If you want to concatenate them into a single DataFrame
    combined_df = pd.concat(dataframes, ignore_index=True)

    #print(combined_df.head())

    combined_df.to_csv(save_dir + '/hereon_' + str(drifter_num) + '_all.csv', index=False)

```
